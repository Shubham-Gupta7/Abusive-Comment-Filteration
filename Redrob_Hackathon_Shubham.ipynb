```python
def normalize_skills(skills):
    skills = skills.lower()
    skills = re.split(r",\s*", skills)
    normalized_skills = []
    for skill in skills:
        if skill in SKILL_ALIASES:
            normalized_skills.append(SKILL_ALIASES[skill])
    return normalized_skills

normalized_resume_dataset = []
for resume in resume_dataset:
    normalized_skills = normalize_skills(resume["skills"])
    normalized_resume_dataset.append({"id": resume["id"], "skills": normalized_skills})
#Print the normalized resume dataset
for i, normalized_resume in enumerate(normalized_resume_dataset):
    original_resume = resume_dataset[i]
    print(f"ID: {normalized_resume['id']}, Name: {original_resume['name']}, Normalized Skills: {normalized_resume['skills']}")
```

```python
def deduplicate_skills(skills):
    return list(set(skills))

deduplicated_resume_dataset = []
for resume in normalized_resume_dataset:
    deduplicated_skills = deduplicate_skills(resume["skills"])
    deduplicated_resume_dataset.append({"id": resume["id"], "skills": deduplicated_skills})

```

```python
# Step 3: Vocabulary Construction
vocabulary = set()
for resume in deduplicated_resume_dataset:
    for skill in resume["skills"]:
        vocabulary.add(skill)
vocabulary = sorted(list(vocabulary))
```

```python
# Step 4: TF-IDF Vector Construction
def compute_tf(skills):
    tf = {}
    if not skills:
        return tf
    for skill in skills:
        if skill not in tf:
            tf[skill] = 0
        tf[skill] += 1
    for skill in tf:
        tf[skill] /= len(skills)
    return tf

# Modified compute_idf to be explicitly global and include smoothing
def compute_idf(vocabulary_list, corpus_dataset):
    idf = {}
    num_docs = len(corpus_dataset)
    for skill_voc in vocabulary_list:
        doc_count = 0
        for doc in corpus_dataset:
            if skill_voc in doc["skills"]:
                doc_count += 1
        # Add 1 to both numerator and denominator for inverse document frequency smoothing (Laplace smoothing)
        idf[skill_voc] = math.log((num_docs + 1) / (doc_count + 1))
    return idf

# Modified compute_tf_idf to ensure dense vectors
def compute_tf_idf(tf_dict, global_idf_map, vocabulary_list):
    tf_idf_vector_dense = {skill: 0.0 for skill in vocabulary_list}
    for skill_in_resume in tf_dict:
        if skill_in_resume in global_idf_map:
            tf_idf_vector_dense[skill_in_resume] = tf_dict[skill_in_resume] * global_idf_map[skill_in_resume]
    return tf_idf_vector_dense

# Calculate global IDF once for the entire corpus
global_idf = compute_idf(vocabulary, deduplicated_resume_dataset)

tf_idf_vectors = []
for resume in deduplicated_resume_dataset:
    tf = compute_tf(resume["skills"])
    tf_idf = compute_tf_idf(tf, global_idf, vocabulary)
    tf_idf_vectors.append(tf_idf)
```

```python
# Step 5: Job Description Vector Construction
def compute_job_description_vector(required_skills, vocabulary):
    vector = {}
    for skill in vocabulary:
        if skill in required_skills:
            vector[skill] = 1
        else:
            vector[skill] = 0
    return vector

job_description_vectors = []
for job_description in JOB_DESCRIPTIONS:
    # Normalize the skills from the job description list by joining them into a string first
    normalized_required_skills = normalize_skills(", ".join(job_description["skills"]))
    vector = compute_job_description_vector(normalized_required_skills, vocabulary)
    job_description_vectors.append(vector)
```

```python
# Step 6: Cosine Similarity Computation
def compute_cosine_similarity(tf_idf_vector, job_description_vector):
    dot_product = 0
    magnitude_tf_idf = 0
    magnitude_job_description = 0

    # Both vectors are now dense (contain all vocabulary skills), so iterating over one is sufficient.
    for skill in tf_idf_vector:
        tf_idf_val = tf_idf_vector[skill]
        jd_val = job_description_vector[skill]

        dot_product += tf_idf_val * jd_val
        magnitude_tf_idf += tf_idf_val ** 2
        magnitude_job_description += jd_val ** 2

    magnitude_tf_idf = math.sqrt(magnitude_tf_idf)
    magnitude_job_description = math.sqrt(magnitude_job_description)

    # Add a small epsilon to the denominator to prevent division by zero in case of extremely small magnitudes
    epsilon = 1e-9
    denominator = (magnitude_tf_idf * magnitude_job_description)

    if denominator < epsilon: # Check if denominator is effectively zero or too small
        return 0

    return dot_product / denominator
```

```python
# Step 7: Ranking and Output
def rank_resumes(job_description_vector, tf_idf_vectors):
    scores = []
    for i, tf_idf_vector in enumerate(tf_idf_vectors):
        score = compute_cosine_similarity(tf_idf_vector, job_description_vector)
        scores.append((i, score))
    scores.sort(key=lambda x: x[1], reverse=True)
    return scores[:3]

for i, job_description_vector in enumerate(job_description_vectors):
    scores = rank_resumes(job_description_vector, tf_idf_vectors)
    print(f"Job Description {i+1}:")
    for score in scores:
        print(f"Resume {score[0]+1}: {score[1]:.2f}")
    print()
```
